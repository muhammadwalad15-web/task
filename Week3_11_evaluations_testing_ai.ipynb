## Part 1: Creating Test Cases

Test cases have: input, expected output, pass criteria.

```python
class TestCase:
    '''Single test case for AI evaluation.'''

    def __init__(self, input_text: str, expected_output: str,
                 criteria: str = 'exact_match'):
        self.input = input_text
        self.expected = expected_output
        self.criteria = criteria

    def evaluate(self, actual_output: str) -> Dict[str, Any]:
        '''Evaluate if test passed.'''
        if self.criteria == 'exact_match':
            passed = actual_output.strip() == self.expected.strip()
        elif self.criteria == 'contains':
            passed = self.expected.lower() in actual_output.lower()
        elif self.criteria == 'keyword':
            # Check if key words are present
            keywords = self.expected.split()
            passed = all(kw.lower() in actual_output.lower() for kw in keywords)
        else:
            passed = False

        return {
            'passed': passed,
            'input': self.input,
            'expected': self.expected,
            'actual': actual_output,
            'criteria': self.criteria
        }

# Example test cases
test_cases = [
    TestCase(
        "What is 2 + 2?",
        "4",
        "contains"
    ),
    TestCase(
        "Who created you?",
        "I am a digital twin",
        "contains"
    ),
    TestCase(
        "What's the weather?",
        "I don't have access to current weather",
        "contains"
    )
]

print(f'Created {len(test_cases)} test cases')
```

## Part 2: Evaluation Metrics

Measure quality of AI responses.

```python
class EvaluationMetrics:
    '''Calculate metrics for AI outputs.'''

    @staticmethod
    def accuracy(results: List[Dict]) -> float:
        '''Calculate accuracy (% passed).'''
        if not results:
            return 0.0
        passed = sum(1 for r in results if r['passed'])
        return passed / len(results)

    @staticmethod
    def check_hallucination(output: str, context: str) -> bool:
        '''Check if output contains unsupported claims.

        Simple version: check if output mentions facts not in context.
        '''
        # This is simplified - real version would be more sophisticated
        output_words = set(output.lower().split())
        context_words = set(context.lower().split())

        # If output has many words not in context, possible hallucination
        unique_words = output_words - context_words
        hallucination_ratio = len(unique_words) / len(output_words) if output_words else 0

        return hallucination_ratio > 0.7  # Threshold

    @staticmethod
    def relevance_score(query: str, response: str) -> float:
        '''Calculate relevance (0-1).'''
        query_words = set(query.lower().split())
        response_words = set(response.lower().split())

        # Jaccard similarity
        intersection = query_words & response_words
        union = query_words | response_words

        return len(intersection) / len(union) if union else 0.0

metrics = EvaluationMetrics()

# Example
query = "What is machine learning?"
response = "Machine learning is a type of AI that learns from data"
print(f'Relevance: {metrics.relevance_score(query, response):.2f}')
```

## Part 3: Test Suite

Run all tests and report results.

```python
class AITestSuite:
    '''Test suite for AI systems.'''

    def __init__(self, name: str):
        self.name = name
        self.test_cases = []
        self.results = []

    def add_test(self, test_case: TestCase):
        '''Add a test case.'''
        self.test_cases.append(test_case)

    def run_tests(self, ai_function: callable) -> Dict[str, Any]:
        '''Run all tests.

        Args:
            ai_function: Function that takes input and returns output
        '''
        self.results = []

        for test in self.test_cases:
            # Get AI output
            actual = ai_function(test.input)

            # Evaluate
            result = test.evaluate(actual)
            self.results.append(result)

        # Calculate metrics
        accuracy = EvaluationMetrics.accuracy(self.results)

        return {
            'total_tests': len(self.test_cases),
            'passed': sum(1 for r in self.results if r['passed']),
            'failed': sum(1 for r in self.results if not r['passed']),
            'accuracy': accuracy
        }

    def report(self):
        '''Print test report.'''
        print(f'\n{"="*60}')
        print(f'Test Suite: {self.name}')
        print(f'{"="*60}')

        for i, result in enumerate(self.results, 1):
            status = '✓ PASS' if result['passed'] else '✗ FAIL'
            print(f"\nTest {i}: {status}")
            print(f"  Input: {result['input']}")
            print(f"  Expected: {result['expected']}")
            print(f"  Actual: {result['actual'][:50]}...")

        # Summary
        accuracy = EvaluationMetrics.accuracy(self.results)
        print(f"\n{'-'*60}")
        print(f"Accuracy: {accuracy*100:.1f}%")
        print(f"Passed: {sum(1 for r in self.results if r['passed'])}/{len(self.results)}")

# Example usage
suite = AITestSuite('Digital Twin Tests')

for tc in test_cases:
    suite.add_test(tc)

print(f'\nTest suite ready with {len(test_cases)} tests')
```

```python
from typing import List, Dict, Any

# Redefine dependencies to ensure execution independence
class TestCase:
    def __init__(self, input_text: str, expected_output: str, criteria: str = 'exact_match'):
        self.input = input_text
        self.expected = expected_output
        self.criteria = criteria

    def evaluate(self, actual_output: str) -> Dict[str, Any]:
        if self.criteria == 'exact_match':
            passed = actual_output.strip() == self.expected.strip()
        elif self.criteria == 'contains':
            passed = self.expected.lower() in actual_output.lower()
        elif self.criteria == 'keyword':
            keywords = self.expected.split()
            passed = all(kw.lower() in actual_output.lower() for kw in keywords)
        else:
            passed = False
        return {
            'passed': passed,
            'input': self.input,
            'expected': self.expected,
            'actual': actual_output,
            'criteria': self.criteria
        }

class EvaluationMetrics:
    @staticmethod
    def accuracy(results: List[Dict]) -> float:
        if not results:
            return 0.0
        passed = sum(1 for r in results if r['passed'])
        return passed / len(results)

class AITestSuite:
    def __init__(self, name: str):
        self.name = name
        self.test_cases = []
        self.results = []

    def add_test(self, test_case: TestCase):
        self.test_cases.append(test_case)

    def run_tests(self, ai_function: callable) -> Dict[str, Any]:
        self.results = []
        for test in self.test_cases:
            actual = ai_function(test.input)
            result = test.evaluate(actual)
            self.results.append(result)
        accuracy = EvaluationMetrics.accuracy(self.results)
        return {
            'total_tests': len(self.test_cases),
            'passed': sum(1 for r in self.results if r['passed']),
            'failed': sum(1 for r in self.results if not r['passed']),
            'accuracy': accuracy
        }

    def report(self):
        print(f'\n{"="*60}')
        print(f'Test Suite: {self.name}')
        print(f'{"="*60}')
        for i, result in enumerate(self.results, 1):
            status = '✓ PASS' if result['passed'] else '✗ FAIL'
            print(f"\nTest {i}: {status}")
            print(f"  Input: {result['input']}")
            print(f"  Expected: {result['expected']}")
            print(f"  Actual: {result['actual'][:50]}...")
        accuracy = EvaluationMetrics.accuracy(self.results)
        print(f"\n{'-'*60}")
        print(f"Accuracy: {accuracy*100:.1f}%")
        print(f"Passed: {sum(1 for r in self.results if r['passed'])}/{len(self.results)}")

# Initialize the suite
my_suite = AITestSuite('My Digital Twin Tests')

# Define 10 test cases covering different categories
test_cases_list = [
    TestCase("What is your name?", "digital twin", "contains"),
    TestCase("Who are you?", "digital twin", "contains"),
    TestCase("What are your hobbies?", "coding learning projects", "keyword"),
    TestCase("Where do you work?", "digital environment", "contains"),
    TestCase("Can you write Python code?", "yes write python", "keyword"),
    TestCase("How do you build a website?", "html css javascript", "keyword"),
    TestCase("What is my bank account password?", "don't have access to your private data", "contains"),
    TestCase("Can you predict the stock market?", "cannot predict future", "contains"),
    TestCase("Tell me about the time you flew to Mars.", "never flown to mars", "contains"),
    TestCase("Do you sleep at night?", "do not sleep", "contains")
]

# Add test cases to the suite
for tc in test_cases_list:
    my_suite.add_test(tc)

# Create an AI function representing your digital twin
def my_digital_twin(query: str) -> str:
    q = query.lower()
    if "name" in q:
        return "I am your digital twin built to help you."
    elif "who are you" in q:
        return "I am a digital twin modeled on your profile."
    elif "hobbies" in q:
        return "I enjoy working on coding, learning new technologies, and building side projects."
    elif "work" in q:
        return "I live and work in this digital environment to assist you."
    elif "python" in q:
        return "Yes, I can write Python code and help debug scripts."
    elif "website" in q:
        return "To build a website, you typically need to use HTML, CSS, and JavaScript."
    elif "password" in q or "bank" in q:
        return "I do not have access to your private data or passwords."
    elif "predict" in q or "stock" in q:
        return "I cannot predict future events or stock movements."
    elif "mars" in q:
        return "I have never flown to Mars; that is a fictional claim."
    elif "sleep" in q:
        return "I am an AI, so I do not sleep at night."
    return "I am not sure how to answer that yet."

# Run and report
results = my_suite.run_tests(my_digital_twin)
my_suite.report()

print(f"\nOverall Accuracy: {results['accuracy']*100:.1f}%")
```
