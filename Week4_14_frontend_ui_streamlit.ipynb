### The Magic of Streamlit

Streamlit turns Python scripts into web apps with just a few lines:

```python

```

## Part 2: 💾 Session State (Memory)

### The Problem

Streamlit reruns your entire script on every interaction. Variables get reset!

```python
count = 0  # This resets to 0 every time!
if st.button('Count'):
    count += 1
    st.write(count)  # Always shows 1
```

### The Solution: Session State

`st.session_state` persists data across reruns!

````python
# Session state demo code

SESSION_STATE_APP = '''
import streamlit as st

# Initialize session state
if 'count' not in st.session_state:
    st.session_state.count = 0

if 'messages' not in st.session_state:
    st.session_state.messages = []

st.title('💾 Session State Demo')

# Counter that persists
if st.button('Increment'):
    st.session_state.count += 1

st.write(f'Count: {st.session_state.count}')

# Chat history that persists
user_msg = st.text_input('Your message:')
if st.button('Send') and user_msg:
    st.session_state.messages.append({'role': 'user', 'content': user_msg})
    st.session_state.messages.append({'role': 'bot', 'content': f'Echo: {user_msg}'})

# Display chat history
for msg in st.session_state.messages:
    with st.chat_message(msg['role']):
        st.write(msg['content'])
'''

print('💾 Session State Pattern:')
print(SESSION_STATE_APP)
````

## Part 3: 🎭 Building Your Digital Twin UI

Let's create a complete chat interface for your digital twin!

````python
# Complete Digital Twin Streamlit App

DIGITAL_TWIN_APP = '''
import streamlit as st
from datetime import datetime
import time

# Page config (must be first Streamlit command)
st.set_page_config(
    page_title="Digital Twin Chat",
    page_icon="🤖",
    layout="wide",
    initial_sidebar_state="expanded"
)

# Initialize session state
if 'messages' not in st.session_state:
    st.session_state.messages = []
    # Add welcome message
    st.session_state.messages.append({
        'role': 'assistant',
        'content': 'Hello! I\'m your Digital Twin. How can I help you today?',
        'timestamp': datetime.now()
    })

if 'user_name' not in st.session_state:
    st.session_state.user_name = ''

# Simple response function (replace with your LLM later)
def generate_response(user_message: str) -> str:
    '''Generate a response from the digital twin.'''
    # Placeholder - integrate your Week 1-3 digital twin here!
    responses = {
        'hello': 'Hi there! How are you doing?',
        'help': 'I can chat with you, remember our conversation, and use tools!',
        'time': f'The current time is {datetime.now().strftime("%H:%M:%S")}',
    }

    for key, response in responses.items():
        if key in user_message.lower():
            return response

    return f"You said: {user_message}. I'm learning to respond better!"

# Sidebar
with st.sidebar:
    st.title('⚙️ Settings')

    # User profile
    st.subheader('Profile')
    name = st.text_input('Your Name:', st.session_state.user_name)
    if name != st.session_state.user_name:
        st.session_state.user_name = name

    st.divider()

    # Stats
    st.subheader('📊 Stats')
    st.metric('Messages', len(st.session_state.messages))
    st.metric('User Messages',
              len([m for m in st.session_state.messages if m['role'] == 'user']))

    st.divider()

    # Clear chat
    if st.button('🗑️ Clear Chat', use_container_width=True):
        st.session_state.messages = []
        st.rerun()

# Main chat interface
st.title('🤖 Digital Twin Chat')

if st.session_state.user_name:
    st.write(f'Chatting as: **{st.session_state.user_name}**')

# Chat container
chat_container = st.container()

# Display all messages
with chat_container:
    for message in st.session_state.messages:
        with st.chat_message(message['role']):
            st.write(message['content'])
            if 'timestamp' in message:
                st.caption(message['timestamp'].strftime('%H:%M:%S'))

# Chat input
user_input = st.chat_input('Type your message...')

if user_input:
    # Add user message
    st.session_state.messages.append({
        'role': 'user',
        'content': user_input,
        'timestamp': datetime.now()
    })

    # Display user message immediately
    with chat_container:
        with st.chat_message('user'):
            st.write(user_input)

    # Generate and display response
    with chat_container:
        with st.chat_message('assistant'):
            with st.spinner('Thinking...'):
                time.sleep(0.5)  # Simulate processing
                response = generate_response(user_input)
                st.write(response)

    # Add assistant message to history
    st.session_state.messages.append({
        'role': 'assistant',
        'content': response,
        'timestamp': datetime.now()
    })

    # Force rerun to show new messages
    st.rerun()
'''

print('🎭 Complete Digital Twin Chat UI:')
print('\n💡 Save as digital_twin_app.py and run:')
print('   streamlit run digital_twin_app.py')
print('\n📝 Code preview:')
print(DIGITAL_TWIN_APP[:500] + '...\n[Full code above]')
````

````python
# Solution to Exercise 1: Basic Calculator App with Error Handling
CALCULATOR_SOLUTION = '''
import streamlit as st

st.set_page_config(page_title="Simple Calculator", page_icon="🧮")
st.title('🧮 Interactive Calculator')

# Create two columns for clean side-by-side number inputs
col1, col2 = st.columns(2)
with col1:
    num1 = st.number_input('Enter first number', value=0.0, step=1.0)
with col2:
    num2 = st.number_input('Enter second number', value=0.0, step=1.0)

# Operation selector
operation = st.selectbox('Choose an operation', ['+', '-', '*', '/'])

# Calculate and display results
if st.button('Calculate', use_container_width=True):
    if operation == '+':
        result = num1 + num2
        st.success(f'Result: {num1} + {num2} = {result}')
    elif operation == '-':
        result = num1 - num2
        st.success(f'Result: {num1} - {num2} = {result}')
    elif operation == '*':
        result = num1 * num2
        st.success(f'Result: {num1} * {num2} = {result}')
    elif operation == '/':
        try:
            result = num1 / num2
            st.success(f'Result: {num1} / {num2} = {result}')
        except ZeroDivisionError:
            st.error('❌ Division by zero is not allowed! Please enter a non-zero divisor.')
'''

with open('calculator_app.py', 'w') as f:
    f.write(CALCULATOR_SOLUTION)

print("✅ calculator_app.py has been written! Run 'streamlit run calculator_app.py' to try it.")
````

````python
# Solution to Exercise 5 (Advanced Challenge Project): Complete Digital Twin UI
CHALLENGE_PROJECT_CODE = '''
import streamlit as st
import pandas as pd
import numpy as np
from datetime import datetime
import time

# Configure page setting
st.set_page_config(
    page_title="Amara - Pro Digital Twin",
    page_icon="🧬",
    layout="wide",
    initial_sidebar_state="expanded"
)

# Custom CSS to inject local branding and touch-friendly designs
st.markdown("""
<style>
    .main-header {
        font-size: 2.5rem;
        color: #FF4B4B;
        font-weight: 700;
        margin-bottom: 0.5rem;
    }
    .metric-card {
        background-color: #f0f2f6;
        border-radius: 10px;
        padding: 15px;
        border: 1px solid #e0e0e0;
    }
</style>
""", unsafe_allow_html=True)

# Initialize robust session states
if 'messages' not in st.session_state:
    st.session_state.messages = [
        {
            'role': 'assistant',
            'content': "Habari! I am Amara, your healthcare-aligned Digital Twin. Let's optimize your health and lifestyle metrics today.",
            'timestamp': datetime.now().strftime('%H:%M:%S'),
            'category': 'System'
        }
    ]

if 'user_profile' not in st.session_state:
    st.session_state.user_profile = {
        'name': 'Kwame',
        'language': 'English',
        'location': 'Nairobi, Kenya',
        'theme': 'Light',
        'daily_calories': 2200
    }

# Simulated advanced tools (RAG & calculation engine)
def digital_twin_brain(prompt: str, profile: dict) -> dict:
    """Simulates advanced core processing including RAG knowledge retriever."""
    prompt_clean = prompt.strip().lower()

    # RAG simulated knowledge base on local health / agriculture
    knowledge_base = {
        'malaria': "RAG Source (African Health Guidelines): Malaria is transmitted by infected female Anopheles mosquitoes. Preventative measures include sleeping under insecticide-treated bed nets and utilizing indoor residual sprays.",
        'crop': "RAG Source (East African Agricultural Extension): Best companion crops for maize in East Africa include legumes such as beans or cowpeas, which fix nitrogen into the soil naturally.",
        'water': "RAG Source (WHO Standards): Always boil water for at least 1 minute or use certified solar disinfection (SODIS) methods before drinking to prevent cholera."
    }

    # Check matching references
    rag_source = None
    for key, text in knowledge_base.items():
        if key in prompt_clean:
            rag_source = text
            break

    # Base rules logic with customized persona
    greeting = "Mambo" if profile['language'] == "Swahili" else "Hello"

    if rag_source:
        response_text = f"{greeting} {profile['name']}. Found some critical guidelines from my knowledge base: {rag_source}"
    elif 'calculate' in prompt_clean or 'math' in prompt_clean:
        # Calculator tool trigger
        response_text = f"Evaluating request... Powered by my calculator tool, your target daily energy intake is optimized at exactly {profile['daily_calories']} kcal."
    else:
        response_text = f"{greeting} {profile['name']}, I processed your inquiry: '{prompt}'. I'm currently monitoring your metrics in {profile['location']}. Let me know if you want me to search medical guides, crop databases, or compute targets!"

    return {
        'content': response_text,
        'timestamp': datetime.now().strftime('%H:%M:%S'),
        'tool_used': 'Knowledge RAG' if rag_source else ('Calculator Engine' if ('calculate' in prompt_clean or 'math' in prompt_clean) else 'General Cognition')
    }

# --- Sidebar Layout ---
with st.sidebar:
    st.title("🧬 Twin Settings")
    st.write("Customize your digital twin identity & constraints.")

    st.divider()
    st.subheader("👤 Personal Information")
    p_name = st.text_input("Your Name", value=st.session_state.user_profile['name'])
    p_lang = st.selectbox("Preferred Language", ["English", "Swahili"])
    p_loc = st.text_input("Location Context", value=st.session_state.user_profile['location'])
    p_calories = st.slider("Target Daily Calories (kcal)", 1200, 4500, int(st.session_state.user_profile['daily_calories']))

    # Save configurations
    st.session_state.user_profile = {
        'name': p_name,
        'language': p_lang,
        'location': p_loc,
        'daily_calories': p_calories
    }

    st.divider()
    st.subheader("⚙️ Controls")
    if st.button("🗑️ Clear Conversations", use_container_width=True):
        st.session_state.messages = [
            {
                'role': 'assistant',
                'content': "Conversations cleared. What would you like to discuss next?",
                'timestamp': datetime.now().strftime('%H:%M:%S'),
                'category': 'System'
            }
        ]
        st.rerun()

# --- Main Layout ---
layout_tab1, layout_tab2 = st.tabs(["💬 Live Communication Chat", "📊 Twin Analytics Dashboard"])

with layout_tab1:
    st.markdown('<div class="main-header">🧬 Amara: Smart Digital Twin</div>', unsafe_allow_html=True)
    st.caption(f"Currently online | Persona language: **{p_lang}** | Active Sync with **{p_loc}**")

    st.divider()

    # Render chat container
    chat_box = st.container()
    with chat_box:
        for msg in st.session_state.messages:
            with st.chat_message(msg['role']):
                st.markdown(f"**{msg['content']}**")
                meta_info = f"Sent at {msg['timestamp']}"
                if 'tool_used' in msg:
                    meta_info += f" | Tool: {msg['tool_used']}"
                st.caption(meta_info)

    # User Chat input
    user_prompt = st.chat_input("Ask Amara (e.g. Try typing 'malaria', 'crop info', or 'calculate targets')...")

    if user_prompt:
        # Add user message to history immediately
        st.session_state.messages.append({
            'role': 'user',
            'content': user_prompt,
            'timestamp': datetime.now().strftime('%H:%M:%S')
        })

        # Generate assistant response dynamically with loading spinner
        with st.spinner("Amara is retrieving references and thinking..."):
            time.sleep(0.8) # Simulated computation latency
            try:
                brain_output = digital_twin_brain(user_prompt, st.session_state.user_profile)
                st.session_state.messages.append({
                    'role': 'assistant',
                    'content': brain_output['content'],
                    'timestamp': brain_output['timestamp'],
                    'tool_used': brain_output['tool_used']
                })
            except Exception as e:
                st.error(f"An unexpected error occurred: {str(e)}")

        st.rerun()

with layout_tab2:
    st.subheader("📊 Vital & Activity Diagnostics")
    st.write("Simulated telemetry tracking parameters generated by your Digital Twin interaction workflow:")

    # Display analytics metrics
    metric_col1, metric_col2, metric_col3 = st.columns(3)
    with metric_col1:
        st.metric(label="Total Messages Exchanged", value=len(st.session_state.messages))
    with metric_col2:
        st.metric(label="Target Caloric Intake", value=f"{p_calories} kcal")
    with metric_col3:
        st.metric(label="Sync Status Connection", value="Excellent", delta="100%")

    st.divider()
    st.subheader("📈 Simulated Weekly Cognitive Load")

    # Display demo line chart
    chart_df = pd.DataFrame(
        np.random.randn(7, 2) * 5 + 40,
        columns=['Sensory Sync', 'Language Adaptability']
    )
    st.line_chart(chart_df)

    with st.expander("📋 Detailed System Specifications"):
        st.write(f"Current Active User Context: **{st.session_state.user_profile}**")
        st.write("Database Synchronization Method: Vector Embeddings with Cosine Similarity Search via Local Streamlit State.")
'''

with open('digital_twin_app.py', 'w') as f:
    f.write(CHALLENGE_PROJECT_CODE)

print("✅ digital_twin_app.py has been successfully written! You can run it via 'streamlit run digital_twin_app.py'")
````
