### 🎯 Exercise 1.2 (Easy): Text Cleaning Pipeline

**Task**: Create a function to clean user input for the digital twin.

**Requirements**:
1. Remove extra whitespace
2. Convert to lowercase
3. Remove punctuation
4. Remove stop words (common words like 'the', 'a', 'is')

````python
# YOUR CODE HERE
def clean_text(text: str, remove_stopwords: bool = True) -> str:
    """
    Clean text for digital twin processing.

    Args:
        text: Input text
        remove_stopwords: Whether to remove common words

    Returns:
        str: Cleaned text
    """
    # Define stopwords
    stopwords = {'the', 'a', 'an', 'and', 'or', 'but', 'is', 'are', 'was', 'were',
                 'in', 'on', 'at', 'to', 'for', 'of', 'with', 'by'}

    # Step 1: Strip whitespace
    cleaned = text.strip()

    # Step 2: Convert to lowercase
    cleaned = cleaned.lower()

    # Step 3: Remove punctuation
    cleaned = cleaned.translate(str.maketrans('', '', string.punctuation))

    # Step 4: Remove stopwords (if enabled)
    if remove_stopwords:
        words = cleaned.split()
        words = [word for word in words if word not in stopwords]
        cleaned = ' '.join(words)

    return cleaned

# Test your function
test_texts = [
    "  I LOVE running in the morning!  ",
    "The weather is perfect for outdoor activities.",
    "Coffee, coding, and creativity are my passions!!!"
]

for text in test_texts:
    cleaned = clean_text(text)
    print(f"Original: '{text}'")
    print(f"Cleaned:  '{cleaned}'")
    print()
````

### 2.1 Regex Basics

````python
# Digital twin conversation with multiple data types
conversation = """
User: My email is john.doe@email.com and I live at 123 Main St.
User: Call me at +1-555-123-4567 or text me!
User: I usually wake up at 6:30am and sleep by 10:45pm.
User: Check out my profile: https://example.com/profile/johndoe
"""

print("Conversation:")
print(conversation)
print("\n" + "="*60)

# Extract emails
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b'
emails = re.findall(email_pattern, conversation)
print(f"\n📧 Emails found: {emails}")

# Extract phone numbers
phone_pattern = r'\+?\d{1,3}[-.]?\d{3}[-.]?\d{3}[-.]?\d{4}'
phones = re.findall(phone_pattern, conversation)
print(f"📱 Phone numbers found: {phones}")

# Extract times
time_pattern = r'\d{1,2}:\d{2}(?:am|pm)'
times = re.findall(time_pattern, conversation)
print(f"🕐 Times found: {times}")

# Extract URLs
url_pattern = r'https?://[^\s]+'
urls = re.findall(url_pattern, conversation)
print(f"🔗 URLs found: {urls}")
````

### 2.2 Common Regex Patterns for Digital Twins

````python
# Pattern library for digital twin data extraction
PATTERNS = {
    'email': r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b',
    'phone': r'\+?\d{1,3}[-.]?\d{3}[-.]?\d{3}[-.]?\d{4}',
    'date': r'\d{4}-\d{2}-\d{2}',
    'time': r'\d{1,2}:\d{2}(?::\d{2})?(?:am|pm)?',
    'url': r'https?://[^\s]+',
    'hashtag': r'#\w+',
    'mention': r'@\w+',
    'number': r'\b\d+(?:\.\d+)?\b',
    'duration': r'\d+(?:h|hr|hours?|m|min|minutes?|s|sec|seconds?)',
}

def extract_patterns(text: str, pattern_name: str) -> List[str]:
    """
    Extract all matches for a given pattern.

    Args:
        text: Input text
        pattern_name: Name of pattern from PATTERNS dict

    Returns:
        list: All matches
    """
    if pattern_name not in PATTERNS:
        raise ValueError(f"Unknown pattern: {pattern_name}")

    pattern = PATTERNS[pattern_name]
    return re.findall(pattern, text, re.IGNORECASE)

# Test with social media-like text
social_text = """
Just finished a 5km run in 30min! 🏃 #running #fitness
@FitnessCoach check out my progress at https://myapp.com/progress
Next goal: 10km in under 1hr. Training starts 2026-01-15 at 6:00am!
"""

print("Text:", social_text)
print("\n" + "="*60)
print("\nExtracted data:")

for pattern_name in ['number', 'duration', 'hashtag', 'mention', 'url', 'date', 'time']:
    matches = extract_patterns(social_text, pattern_name)
    if matches:
        print(f"  {pattern_name.title()}: {matches}")
````

### 🎯 Exercise 2.1 (Intermediate): Build a Personal Data Extractor

**Task**: Create a comprehensive data extraction system for digital twin profiles.

**Requirements**:
1. Extract all personal identifiers (email, phone, social handles)
2. Extract temporal information (dates, times, durations)
3. Extract preferences and activities from hashtags
4. Return structured data (dictionary)

````python
# YOUR CODE HERE
def extract_profile_data(text: str) -> Dict[str, any]:
    """
    Extract structured data from user text.

    Args:
        text: User profile or conversation text

    Returns:
        dict: Extracted data organized by category
    """
    profile = {
        'contact': {},
        'temporal': {},
        'interests': [],
        'metrics': {},
    }

    # Patterns dictionary helper
    email_matches = re.findall(PATTERNS['email'], text, re.IGNORECASE)
    phone_matches = re.findall(PATTERNS['phone'], text, re.IGNORECASE)
    date_matches = re.findall(PATTERNS['date'], text, re.IGNORECASE)
    time_matches = re.findall(PATTERNS['time'], text, re.IGNORECASE)
    hashtag_matches = re.findall(PATTERNS['hashtag'], text, re.IGNORECASE)
    duration_matches = re.findall(PATTERNS['duration'], text, re.IGNORECASE)
    url_matches = re.findall(PATTERNS['url'], text, re.IGNORECASE)
    mention_matches = re.findall(PATTERNS['mention'], text, re.IGNORECASE)

    # Extract contact information
    if email_matches:
        profile['contact']['emails'] = email_matches
    if phone_matches:
        profile['contact']['phones'] = phone_matches
    if url_matches:
        profile['contact']['urls'] = url_matches
    if mention_matches:
        profile['contact']['mentions'] = mention_matches

    # Extract temporal data
    if date_matches:
        profile['temporal']['dates'] = date_matches
    if time_matches:
        profile['temporal']['times'] = time_matches
    if duration_matches:
        profile['temporal']['durations'] = duration_matches

    # Extract interests from hashtags
    if hashtag_matches:
        profile['interests'] = [h.replace('#', '') for h in hashtag_matches]

    # Extract numerical metrics
    # Finds standalone metrics e.g., "5km", "50min"
    metrics_pattern = r'\b\d+(?:km|min|hr|hours?|m)\b'
    metrics_matches = re.findall(metrics_pattern, text, re.IGNORECASE)
    if metrics_matches:
        profile['metrics']['values'] = metrics_matches

    return profile

# Test with comprehensive profile
test_profile = """
Hi! I'm Sarah (sarah.runner@email.com). Reach me at +1-555-234-5678.
I run 5km daily at 6:30am and love #running #fitness #health.
My best time: 25min for 5km. Goal: 10km in 50min by 2026-03-01.
Follow my journey: https://strava.com/sarah_runs and @SarahRuns on Twitter!
"""

extracted = extract_profile_data(test_profile)
print(json.dumps(extracted, indent=2))
````

## Part 3: Text Chunking for RAG Systems

**Critical for RAG**: Breaking documents into optimal chunks for retrieval.

### 3.1 Simple Chunking Strategies

````python
# Sample document about digital twin
long_document = """
A digital twin is a virtual representation of a physical object or system.
Digital twins are used across various industries. In healthcare, they model patient health.
In manufacturing, they simulate production processes. Smart cities use digital twins for urban planning.

The technology relies on real-time data from sensors. This data feeds into simulation models.
Machine learning algorithms analyze patterns. Predictions help optimize operations.

Benefits include improved decision-making. Companies reduce costs through predictive maintenance.
Healthcare providers personalize treatments. City planners optimize resource allocation.
"""

def chunk_by_sentences(text: str, sentences_per_chunk: int = 2) -> List[str]:
    """
    Chunk text by sentences.

    Args:
        text: Input text
        sentences_per_chunk: Number of sentences per chunk

    Returns:
        list: Text chunks
    """
    # Simple sentence splitting (not perfect, but educational)
    sentences = re.split(r'(?<=[.!?])\s+', text.strip())

    chunks = []
    for i in range(0, len(sentences), sentences_per_chunk):
        chunk = ' '.join(sentences[i:i+sentences_per_chunk])
        chunks.append(chunk)

    return chunks

def chunk_by_words(text: str, words_per_chunk: int = 50, overlap: int = 10) -> List[str]:
    """
    Chunk text by word count with overlap.

    Args:
        text: Input text
        words_per_chunk: Target words per chunk
        overlap: Number of overlapping words

    Returns:
        list: Text chunks
    """
    words = text.split()
    chunks = []

    for i in range(0, len(words), words_per_chunk - overlap):
        chunk = ' '.join(words[i:i+words_per_chunk])
        chunks.append(chunk)

        if i + words_per_chunk >= len(words):
            break

    return chunks

# Test chunking strategies
print("Original document length:", len(long_document), "characters")
print("\n" + "="*60)

# Sentence-based chunking
sentence_chunks = chunk_by_sentences(long_document, sentences_per_chunk=3)
print(f"\nSentence-based chunks ({len(sentence_chunks)} chunks):")
for i, chunk in enumerate(sentence_chunks, 1):
    print(f"\nChunk {i} ({len(chunk)} chars):")
    print(chunk[:100] + "..." if len(chunk) > 100 else chunk)

# Word-based chunking with overlap
print("\n" + "="*60)
word_chunks = chunk_by_words(long_document, words_per_chunk=30, overlap=5)
print(f"\nWord-based chunks ({len(word_chunks)} chunks, 30 words/chunk, 5 word overlap):")
for i, chunk in enumerate(word_chunks, 1):
    word_count = len(chunk.split())
    print(f"\nChunk {i} ({word_count} words):")
    print(chunk[:100] + "..." if len(chunk) > 100 else chunk)
````

### 3.2 Semantic Chunking (Advanced)

````python
def chunk_by_paragraphs(text: str) -> List[str]:
    """
    Chunk text by paragraphs (natural semantic boundaries).

    Args:
        text: Input text

    Returns:
        list: Paragraph chunks
    """
    # Split on double newlines
    paragraphs = re.split(r'\n\s*\n', text.strip())
    # Clean each paragraph
    chunks = [p.strip() for p in paragraphs if p.strip()]
    return chunks

def smart_chunk(text: str,
                max_chunk_size: int = 200,
                overlap: int = 20) -> List[Dict[str, any]]:
    """
    Smart chunking that respects semantic boundaries.

    Args:
        text: Input text
        max_chunk_size: Maximum words per chunk
        overlap: Overlap words between chunks

    Returns:
        list: Chunks with metadata
    """
    # First, try to split by paragraphs
    paragraphs = chunk_by_paragraphs(text)

    chunks = []
    current_chunk = ""

    for para in paragraphs:
        para_words = len(para.split())
        current_words = len(current_chunk.split())

        # If adding paragraph exceeds limit, save current and start new
        if current_words + para_words > max_chunk_size and current_chunk:
            chunks.append({
                'text': current_chunk.strip(),
                'word_count': len(current_chunk.split()),
                'char_count': len(current_chunk)
            })

            # Start new chunk with overlap from previous
            words = current_chunk.split()
            overlap_text = ' '.join(words[-overlap:]) if len(words) > overlap else current_chunk
            current_chunk = overlap_text + ' ' + para
        else:
            current_chunk = current_chunk + '\n' + para if current_chunk else para

    # Add final chunk
    if current_chunk:
        chunks.append({
            'text': current_chunk.strip(),
            'word_count': len(current_chunk.split()),
            'char_count': len(current_chunk)
        })

    return chunks

# Test smart chunking
smart_chunks = smart_chunk(long_document, max_chunk_size=50, overlap=10)

print("Smart Chunking Results:")
print(f"Total chunks: {len(smart_chunks)}\n")

for i, chunk in enumerate(smart_chunks, 1):
    print(f"Chunk {i}:")
    print(f"  Words: {chunk['word_count']}, Chars: {chunk['char_count']}")
    print(f"  Text preview: {chunk['text'][:100]}...")
    print()
````

### 🎯 Exercise 3.1 (Advanced): Build a Chunking System

**Task**: Create a configurable chunking system for the digital twin's knowledge base.

**Requirements**:
1. Support multiple chunking strategies (sentence, word, paragraph, smart)
2. Add metadata to each chunk (position, source, timestamp)
3. Validate chunk quality (no orphaned sentences, reasonable size)
4. Create visualization of chunk distribution

````python
import re
import time
from typing import List, Dict, Any
import matplotlib.pyplot as plt

class ChunkingSystem:
    """
    Flexible chunking system for RAG preparation.
    """

    def __init__(self, strategy: str = 'smart'):
        self.strategy = strategy
        self.chunks = []

    def chunk(self, text: str, **kwargs) -> List[Dict[str, Any]]:
        """
        Chunk text using selected strategy.
        """
        strategy = kwargs.get('strategy', self.strategy)
        if strategy == 'paragraph':
            paragraphs = re.split(r'\n\s*\n', text.strip())
            raw_chunks = [p.strip() for p in paragraphs if p.strip()]
        elif strategy == 'sentence':
            sentences_per_chunk = kwargs.get('sentences_per_chunk', 2)
            sentences = re.split(r'(?<=[.!?])\s+', text.strip())
            raw_chunks = []
            for i in range(0, len(sentences), sentences_per_chunk):
                raw_chunks.append(' '.join(sentences[i:i+sentences_per_chunk]))
        elif strategy == 'word':
            words_per_chunk = kwargs.get('words_per_chunk', 50)
            overlap = kwargs.get('overlap', 10)
            words = text.split()
            raw_chunks = []
            for i in range(0, len(words), words_per_chunk - overlap):
                raw_chunks.append(' '.join(words[i:i+words_per_chunk]))
                if i + words_per_chunk >= len(words):
                    break
        else: # smart
            max_chunk_size = kwargs.get('max_chunk_size', 50)
            overlap = kwargs.get('overlap', 10)
            paragraphs = [p.strip() for p in re.split(r'\n\s*\n', text.strip()) if p.strip()]
            raw_chunks = []
            current_chunk = ""
            for para in paragraphs:
                para_words = len(para.split())
                current_words = len(current_chunk.split())
                if current_words + para_words > max_chunk_size and current_chunk:
                    raw_chunks.append(current_chunk.strip())
                    words = current_chunk.split()
                    overlap_text = ' '.join(words[-overlap:]) if len(words) > overlap else current_chunk
                    current_chunk = overlap_text + ' ' + para
                else:
                    current_chunk = current_chunk + '\n' + para if current_chunk else para
            if current_chunk:
                raw_chunks.append(current_chunk.strip())

        self.chunks = [{'text': rc, 'word_count': len(rc.split()), 'char_count': len(rc)} for rc in raw_chunks]
        return self.chunks

    def add_metadata(self, chunks: List[Dict], source: str = None) -> List[Dict]:
        """
        Add metadata to chunks.
        """
        timestamp = time.strftime("%Y-%m-%dT%H:%M:%SZ", time.gmtime())
        for i, chunk in enumerate(chunks):
            chunk['metadata'] = {
                'position': i + 1,
                'total_chunks': len(chunks),
                'source': source or 'unknown',
                'timestamp': timestamp
            }
        return chunks

    def validate_chunks(self, chunks: List[Dict]) -> Dict[str, Any]:
        """
        Validate chunk quality.
        """
        total = len(chunks)
        if total == 0:
            return {'status': 'empty'}

        word_counts = [c['word_count'] for c in chunks]
        avg_words = sum(word_counts) / total
        too_small = sum(1 for w in word_counts if w < 5)
        too_large = sum(1 for w in word_counts if w > 150)

        return {
            'status': 'valid' if (too_small == 0 and too_large == 0) else 'review_needed',
            'total_chunks': total,
            'average_word_count': avg_words,
            'chunks_too_small': too_small,
            'chunks_too_large': too_large
        }

    def visualize_distribution(self, chunks: List[Dict]):
        """
        Visualize chunk size distribution.
        """
        word_counts = [c['word_count'] for c in chunks]
        fig, ax = plt.subplots(figsize=(8, 4))
        ax.bar(range(1, len(chunks) + 1), word_counts, color='skyblue', edgecolor='black')
        ax.set_title('Chunk Word Count Distribution')
        ax.set_xlabel('Chunk Index')
        ax.set_ylabel('Word Count')
        plt.show()

# Test your system
chunker = ChunkingSystem(strategy='smart')
raw_chunks = chunker.chunk(long_document, max_chunk_size=40, overlap=5)
chunks_with_meta = chunker.add_metadata(raw_chunks, source='digital_twin_doc.txt')
validation_report = chunker.validate_chunks(chunks_with_meta)

print("Validation Report:\n", json.dumps(validation_report, indent=2))
chunker.visualize_distribution(chunks_with_meta)
````

## Part 4: Real-World Text Processing Pipelines

Putting it all together for production systems.

### 4.1 Complete Processing Pipeline

````python
class TextProcessor:
    """
    Complete text processing pipeline for digital twin.
    """

    def __init__(self):
        self.processing_history = []

    def process(self, text: str,
                clean: bool = True,
                extract_data: bool = True,
                chunk: bool = False,
                chunk_size: int = 100) -> Dict[str, any]:
        """
        Process text through complete pipeline.

        Args:
            text: Input text
            clean: Apply cleaning
            extract_data: Extract structured data
            chunk: Chunk the text
            chunk_size: Size for chunking

        Returns:
            dict: Processed results
        """
        result = {
            'original': text,
            'original_stats': self._get_stats(text),
            'processed': text,
            'extracted_data': {},
            'chunks': [],
            'pipeline_steps': []
        }

        # Step 1: Clean
        if clean:
            result['processed'] = self._clean(result['processed'])
            result['pipeline_steps'].append('cleaned')

        # Step 2: Extract structured data
        if extract_data:
            result['extracted_data'] = self._extract(result['processed'])
            result['pipeline_steps'].append('extracted')

        # Step 3: Chunk
        if chunk:
            result['chunks'] = self._chunk(result['processed'], chunk_size)
            result['pipeline_steps'].append('chunked')

        result['processed_stats'] = self._get_stats(result['processed'])

        self.processing_history.append(result)
        return result

    def _clean(self, text: str) -> str:
        """Clean text."""
        # Remove extra whitespace
        text = re.sub(r'\s+', ' ', text)
        text = text.strip()
        return text

    def _extract(self, text: str) -> Dict[str, any]:
        """Extract structured data."""
        data = {}
        for pattern_name, pattern in PATTERNS.items():
            matches = re.findall(pattern, text, re.IGNORECASE)
            if matches:
                data[pattern_name] = matches
        return data

    def _chunk(self, text: str, chunk_size: int) -> List[Dict]:
        """Chunk text."""
        return smart_chunk(text, max_chunk_size=chunk_size)

    def _get_stats(self, text: str) -> Dict[str, int]:
        """Get text statistics."""
        return {
            'chars': len(text),
            'words': len(text.split()),
            'lines': len(text.split('\n')),
        }

# Test the complete pipeline
processor = TextProcessor()

sample_text = """
Hi! I'm Alex (alex@email.com). I love running and have been tracking my progress.
Yesterday I ran 5km in 28 minutes at 6:30am. My goal is to complete a marathon (42km) by 2026-06-15.

I also enjoy coding in Python and building ML models. Check out my GitHub: https://github.com/alexruns

Daily routine:
- Wake up at 5:45am
- Run for 45-60min
- Work from 9:00am to 5:00pm
- Code on side projects from 8:00pm to 10:00pm

#running #fitness #coding #python #AI
"""

result = processor.process(
    sample_text,
    clean=True,
    extract_data=True,
    chunk=True,
    chunk_size=50
)

print("Processing Results:")
print(f"Pipeline steps: {result['pipeline_steps']}")
print(f"\nOriginal stats: {result['original_stats']}")
print(f"Processed stats: {result['processed_stats']}")
print(f"\nExtracted data:")
for key, values in result['extracted_data'].items():
    print(f"  {key}: {values}")
print(f"\nChunks created: {len(result['chunks'])}")
````

3. Cleans and normalizes text
4. Chunks optimally for RAG
5. Generates metadata for each chunk
6. Creates search indices
7. Provides quality metrics

**Bonus**: Add support for handling tables, code blocks, and special formatting!

````python
import re
import time
import json
from typing import List, Dict, Any
from collections import defaultdict
import math

class DocumentPreprocessor:
    """
    A complete Document Preprocessing & Indexing system for a Digital Twin Knowledge Base
    with built-in support for Markdown files.
    """
    def __init__(self, chunk_strategy: str = 'smart'):
        self.chunker = ChunkingSystem(strategy=chunk_strategy)
        self.documents_store = {}
        self.chunks_store = []
        self.index = defaultdict(list)

    def ingest_document(self, doc_id: str, content: str, doc_type: str = "txt") -> Dict[str, Any]:
        """
        Ingests, cleans, extracts structured data, parses markdown, chunks, and indexes a document.
        """
        # 1. Clean and normalize
        cleaned_content = self._clean_and_normalize(content)

        # 2. Extract structured data
        extracted_data = extract_profile_data(content)

        # 3. Create Chunks
        raw_chunks = self.chunker.chunk(cleaned_content, max_chunk_size=40, overlap=5)
        chunks_with_metadata = self.chunker.add_metadata(raw_chunks, source=doc_id)

        # Keep track of custom structured info inside chunks
        for chunk in chunks_with_metadata:
            chunk['metadata']['doc_type'] = doc_type
            chunk['extracted_data'] = extracted_data

        # 4. Update memory stores
        self.documents_store[doc_id] = {
            'raw': content,
            'cleaned': cleaned_content,
            'doc_type': doc_type,
            'extracted': extracted_data
        }

        # If doc_type is markdown, parse specific markdown elements
        if doc_type == "markdown":
            self.documents_store[doc_id]['markdown_elements'] = self.parse_markdown_elements(content)

        start_index = len(self.chunks_store)
        self.chunks_store.extend(chunks_with_metadata)

        # 5. Build inverted search index
        self._build_index_for_chunks(chunks_with_metadata, start_index)

        # 6. Quality validation metric
        report = self.chunker.validate_chunks(chunks_with_metadata)

        return {
            'doc_id': doc_id,
            'num_chunks': len(chunks_with_metadata),
            'validation': report
        }

    def parse_markdown_elements(self, text: str) -> Dict[str, Any]:
        """
        Parses markdown text to extract headers, lists, code blocks, and bold text.
        """
        elements = {
            'headers': [],
            'lists': [],
            'code_blocks': [],
            'bold_text': []
        }

        # Extract Headers
        headers = re.findall(r'^(#{1,6})\s+(.+)$', text, re.MULTILINE)
        for h_level, h_text in headers:
            elements['headers'].append({
                'level': len(h_level),
                'text': h_text.strip()
            })

        # Extract Unordered Lists
        lists = re.findall(r'^\s*[-*+]\s+(.+)$', text, re.MULTILINE)
        elements['lists'] = [item.strip() for item in lists]

        # Extract Code Blocks
        code_blocks = re.findall(r'```(\w+)?\n(.*?)\n```', text, re.DOTALL)
        for lang, code_content in code_blocks:
            elements['code_blocks'].append({
                'language': lang or 'text',
                'code': code_content.strip()
            })

        # Extract Bold Text
        bold_items = re.findall(r'\*\*(.*?)\*\*', text)
        elements['bold_text'] = [item.strip() for item in bold_items if item.strip()]

        return elements

    def _clean_and_normalize(self, text: str) -> str:
        # Preserve structures like markdown headers but clean redundant spacings
        cleaned = re.sub(r'\n{3,}', '\n\n', text)
        return cleaned.strip()

    def _build_index_for_chunks(self, chunks: List[Dict], start_idx: int):
        for i, chunk in enumerate(chunks):
            global_idx = start_idx + i
            # Tokenize
            words = clean_text(chunk['text'], remove_stopwords=True).split()
            for word in set(words):
                self.index[word].append(global_idx)

    def search(self, query: str) -> List[Dict[str, Any]]:
        """
        Simple search retrieval using the inverted index.
        """
        query_words = clean_text(query, remove_stopwords=True).split()
        match_counts = defaultdict(int)

        for word in query_words:
            if word in self.index:
                for idx in self.index[word]:
                    match_counts[idx] += 1

        # Sort chunks by relevance matches
        sorted_indices = sorted(match_counts.keys(), key=lambda idx: match_counts[idx], reverse=True)
        return [self.chunks_store[idx] for idx in sorted_indices[:3]]

# Instantiate and run our integrated Digital Twin Document Preprocessor challenge
preprocessor = DocumentPreprocessor(chunk_strategy='smart')

# Document with Markdown, metrics, emails, dates and tags
doc_1 = """
# Digital Twin Plan - Workout Routines
Email contact: coach.mike@twinlabs.com. Date of creation: 2026-02-12.

Today's workout session involves running for 45min at a steady pace.
Focus on maintaining a high heart rate.

Ensure to log all metrics inside #fitness app! Use @CoachMike to notify the mentor.
"""

ingestion_report = preprocessor.ingest_document("workout_plan_01", doc_1, doc_type="markdown")
print("Ingestion & Quality Validation:", json.dumps(ingestion_report, indent=2))

# Query the system
search_results = preprocessor.search("running metrics")
print("\n--- Search Query: 'running metrics' ---")
for idx, result in enumerate(search_results, 1):
    print(f"Match {idx} (Doc: {result['metadata']['source']}):")
    print(f"  Text: {result['text']}")
    print(f"  Extracted Entities: {result['extracted_data']}\n")
````

````python
def parse_markdown_elements(text: str) -> Dict[str, Any]:
    """
    Parses markdown text to extract structured elements like headers, lists, and code blocks.
    """
    elements = {
        'headers': [],
        'lists': [],
        'code_blocks': [],
        'bold_text': []
    }

    # Extract Headers (e.g., # Header, ## Header)
    headers = re.findall(r'^(#{1,6})\s+(.+)$', text, re.MULTILINE)
    for h_level, h_text in headers:
        elements['headers'].append({
            'level': len(h_level),
            'text': h_text.strip()
        })

    # Extract Unordered Lists
    lists = re.findall(r'^\s*[-*+]\s+(.+)$', text, re.MULTILINE)
    elements['lists'] = [item.strip() for item in lists]

    # Extract Code Blocks
    code_blocks = re.findall(r'```(\w+)?\n(.*?)\n```', text, re.DOTALL)
    for lang, code_content in code_blocks:
        elements['code_blocks'].append({
            'language': lang or 'text',
            'code': code_content.strip()
        })

    # Extract Bold Text
    bold_items = re.findall(r'\*\*(.*?)\*\*', text)
    elements['bold_text'] = [item.strip() for item in bold_items if item.strip()]

    return elements

# Extend our DocumentPreprocessor with markdown support
class MarkdownDocumentPreprocessor(DocumentPreprocessor):
    def ingest_document(self, doc_id: str, content: str, doc_type: str = "txt") -> Dict[str, Any]:
        report = super().ingest_document(doc_id, content, doc_type)
        if doc_type == "markdown":
            parsed_elements = parse_markdown_elements(content)
            self.documents_store[doc_id]['markdown_elements'] = parsed_elements
            report['markdown_elements_count'] = {
                k: len(v) for k, v in parsed_elements.items()
            }
            # Add quality metrics details of markdown structure
            report['validation']['markdown_specific_metrics'] = {
                'has_headers': len(parsed_elements['headers']) > 0,
                'has_code_blocks': len(parsed_elements['code_blocks']) > 0,
                'list_items_count': len(parsed_elements['lists'])
            }
        return report

# Instantiate and run testing with sample document and explicit chunk-level validation
md_preprocessor = MarkdownDocumentPreprocessor(chunk_strategy='smart')
sample_md_doc = """
# Digital Twin Lab Guide
Welcome to the official **Digital Twin Development Guide**.

## Quick Start Steps
- Create a Google Colab notebook
- Import pattern matching utilities
- Process unstructured datasets

## Example Code Block
```python
print("Initialization complete!")
```
"""

md_report = md_preprocessor.ingest_document(
    doc_id="dt_guide_01",
    content=sample_md_doc,
    doc_type="markdown"
)

print("Markdown Ingestion & Quality Metrics Report:\n", json.dumps(md_report, indent=2))
print("\nParsed Markdown Elements Details:\n", json.dumps(md_preprocessor.documents_store['dt_guide_01']['markdown_elements'], indent=2))
````

````python
def save_chunks_to_json(chunks: List[Dict[str, Any]], filepath: str) -> None:
    """
    Saves the processed chunks list directly into a formatted JSON file.
    """
    try:
        with open(filepath, 'w', encoding='utf-8') as f:
            json.dump(chunks, f, ensure_ascii=False, indent=2)
        print(f"✅ Successfully saved {len(chunks)} chunks to {filepath}!")
    except Exception as e:
        print(f"❌ Failed to save chunks: {e}")

# Let's save the chunks we processed from the sample markdown guide
save_path = "/content/processed_chunks.json"
save_chunks_to_json(md_preprocessor.chunks_store, save_path)

# Verify by reading back a preview of the saved JSON file
with open(save_path, 'r') as f:
    saved_data = json.load(f)
    print(f"\nPreview of the first saved chunk from JSON file:\n")
    print(json.dumps(saved_data[0], indent=2))
````

````python
import pandas as pd

def load_chunks_to_dataframe(filepath: str) -> pd.DataFrame:
    """
    Loads a JSON file containing processed chunks into a pandas DataFrame.
    """
    try:
        df = pd.read_json(filepath)
        # Normalize nested dictionaries (like metadata and extracted_data) into flat columns
        df_meta = pd.json_normalize(df['metadata'])
        df_extracted = pd.json_normalize(df['extracted_data'])

        # Drop the original nested columns and join the flattened ones
        df_flat = df.drop(columns=['metadata', 'extracted_data']).join(df_meta).join(df_extracted)
        print(f"✅ Successfully loaded and flattened {len(df_flat)} chunks into a DataFrame!")
        return df_flat
    except Exception as e:
        print(f"❌ Failed to load DataFrame: {e}")
        return pd.DataFrame()

# Load the processed chunks and display the first few rows
chunks_df = load_chunks_to_dataframe(save_path)
if not chunks_df.empty:
    display(chunks_df.head())
````
